```python
import requests
from bs4 import BeautifulSoup
from pprint import pprint

url = "https://webscraper.io/"

response = requests.get(url, timeout=15)
response.raise_for_status()

soup = BeautifulSoup(response.text, "lxml")

footer = soup.find("footer")

if not footer:
    footer = soup.select_one(".footer, #footer, [class*='footer']")

if footer:
    footer_links = {}

    for link in footer.find_all("a", href=True):
        name = link.get_text(" ", strip=True)
        address = link["href"]

        if address.startswith("/"):
            address = url.rstrip("/") + address
        elif not address.startswith("http"):
            address = url.rstrip("/") + "/" + address

        if name:
            footer_links[name] = address

    print("\nFOOTER")
    print("-" * 50)
    pprint(footer_links, sort_dicts=False, width=100)

    with open("footer.html", "w", encoding="utf-8") as file:
        file.write(str(footer))

    print("\nFooter HTML saved to footer.html")

else:
    print("Could not find the footer in the downloaded HTML.")
```
